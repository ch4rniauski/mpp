# Практическая работа №4
Операции с директориями и файлами в Google Colab

```python
import os
import shutil
from google.colab import files
```

## 1. Создание директории project

```python
os.makedirs('project', exist_ok=True)
print("Директория 'project' создана.")
```

## 2. Переход в директорию project

```python
%cd project
print("Текущая директория:", os.getcwd())
```

## 3. Создание файла data.txt

```python
with open('data.txt', 'w') as file:
    file.write('Первая строка\n')
    file.write('Вторая строка\n')
    file.write('Третья строка\n')
print("Файл 'data.txt' создан и записан.")
```

## 4. Чтение содержимого файла

```python
with open('data.txt', 'r') as file:
    content = file.read()
print("Содержимое файла:")
print(content)
```

## 5. Создание копии data_backup.txt

```python
shutil.copy('data.txt', 'data_backup.txt')
print("Создана копия 'data_backup.txt'.")
```

## 6. Удаление оригинала data.txt

```python
os.remove('data.txt')
print("Файл 'data.txt' удалён.")
```

## 7. Переименование data_backup.txt обратно в data.txt

```python
os.rename('data_backup.txt', 'data.txt')
print("Файл переименован в 'data.txt'.")
```

## 8. Загрузка файла с компьютера в Colab

```python
uploaded = files.upload()
print("Загруженные файлы:", list(uploaded.keys()))
```

## 9. Запись в data.txt содержимого папки /content/

```python
with open('data.txt', 'w') as file:
    for dirpath, _, filenames in os.walk('/content/'):
        for filename in filenames:
            file.write(os.path.join(dirpath, filename) + '\n')
print("Содержимое /content/ записано в 'data.txt'.")

with open('data.txt', 'r') as file:
    print(file.read())
```

## 10. Скачивание data.txt на компьютер

```python
files.download('data.txt')
print("Файл 'data.txt' скачан.")
```
